### VIZUALITZAR

```python
with Session() as session:
    stmt = select(User)
    users = session.scalars(stmt).all()

    for user in users:
        print(f"Name: {user.user_name}")
```

### ORDER BY

```python
with Session() as session:
    stmt = select(User).order_by(User.user_name) # Amb .desc() despres de .name ho ordenem descendentment
    users = session.scalars(stmt).all()

    for user in users:
        print(f"Name: {user.user_name}")
```

### CONSULTA WHERE i AND

```python
with Session() as session:
    stmt = select(User).where(and_(User.alias == "Lluc", User.id == 15))
    users = session.scalars(stmt).all()

    for user in users:
        print(f"Name: {user.user_name}")
```

### CONSULTA WHERE i AND

```python
with Session() as session:
    stmt = select(User).where(or_(User.alias == "Lluc", User.id == 1))
    users = session.scalars(stmt).all()

    for user in users:
        print(f"Name: {user.user_name}")
```

### GET ONE

```python
with Session() as session:
    user = session.query(User).first()
    if user:
        print(user.user_name)
```

```python
with Session() as session:
    name = session.scalar(select(User))

    if name:
        print(f"Nom: {name.user_name}")
```

### GET BY ID

```python
with Session() as session:
    name = session.get(User, 15) # Agafa per id

    if name:
        print(f"Nom: {name.user_name}")
```

### PAGINACIÓ

```python
with Session() as session:
    page_size = 5
    page = 1

    stmt = (
        select(Recipe)
        .order_by(Recipe.name)
        .limit(page_size)
        .offset((page - 1) * page_size) #  li diu a la BD quants registres ha de saltar abans de començar a retornar dades.
    ) # query de l'entitat per ordenar i limita les dades a mostrar
    recipes = session.scalars(stmt).all() # retorna el resultat

    for i in recipes:
        print(i.name)
```
