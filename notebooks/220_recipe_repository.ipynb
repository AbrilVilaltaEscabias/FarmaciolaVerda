### SPECIFIC GET OR FILTER QUERIES

#### GET BY NAME

```python
with UnitOfWork(Session) as uow:

    infusio_de_camamilla = uow.recipes.get_by_name("Infusió de camamilla")

    for i in infusio_de_camamilla:
        print(f"""
Name: {i.name}
Step by step: {i.step_by_step}
        """)
```

#### GET BY CATEGORY

```python
with UnitOfWork(Session) as uow:

    receptes = uow.recipes.get_by_category("Digestiva")

    for r in receptes:
        print(f"""
Name: {r.name}
Step by step: {r.step_by_step}
Categories: {[c.name for c in r.categories]}
        """)
```

### CONCRETE DOMAIN OPERATIONS

#### ADD INGREDIENT

```python
with UnitOfWork(Session) as uow:


    uow.recipes.add_ingredient(1, 9, 3)
    ingredient = uow.ingredients.get(9)
    recipe = uow.recipes.get(1)
    uow.commit()  
    print(f"{ingredient.common_name} added to {recipe.name}'s ingredients list")
```
